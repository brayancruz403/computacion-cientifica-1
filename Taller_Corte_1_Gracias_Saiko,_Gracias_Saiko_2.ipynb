1. Realiza un algoritmo que transforme numeros decimales a numeros binarios

```python
n = input("ingrese un numero para convertir en binario: ")
n = int(n)
n_binario = []

if n == 1:
 print("su numero en binario es: 1")
elif n == 0:
 print("su numero en binario es: 0")
else:
  while n > 0:
    residuo = int(n % 2)
    n_binario.append(residuo)
    n = int(n / 2)

    n_binario.reverse()

print("su numero en binario es:", n_binario)
```

2. Realiza un algoritmo que transforme numeros binarios a numeros decimales

```python
n = input("Ingrese un número binario para convertir en decimal: ")
lista_binaria = list(n)

decimal = 0
longitud = len(lista_binaria)

for i in range(longitud):
    bit = int(lista_binaria[i])
    potencia = longitud - 1 - i
    decimal += bit * (2 ** potencia)

print("Su número en decimal es:", decimal)
```

3. Realiza un algoritmo que transforme numeros romanos a numeros decimales

```python
romanos = { "I": 1, "V": 5, "X": 10, "L": 50, "C": 100, "D": 500, "M": 1000}

texto_romano = input("Ingrese un número romano: ")
texto_romano = texto_romano.upper()

decimal = 0
longitud = len(texto_romano)

for i in range(longitud):
    if i > 0 and romanos[texto_romano[i]] > romanos[texto_romano[i - 1]]:
        decimal += romanos[texto_romano[i]] - 2 * romanos[texto_romano[i - 1]]
    else:
        decimal += romanos[texto_romano[i]]

print("Su número en decimal es:", decimal)

```

4. Realiza un algoritmo que transforme numeros decimales a numeros romanos

```python
valores = [ (1000, "M"), (900, "CM"), (500, "D"), (400, "CD"), (100, "C"), (90, "XC"), (50, "L"), (40, "XL"), (10, "X"), (9, "IX"), (5, "V"), (4, "IV"), (1, "I")]

num = int(input("Ingrese un número entero positivo: "))

resultado = ""
for valor, simbolo in valores:
    while num >= valor:
        resultado += simbolo
        num -= valor

print("Su número en romano es:", resultado)
```

5. Realiza un algoritmo que valide si una frase es un palindromo

```python
def es_palindromo(frase):
    frase = frase.lower().replace(" ", "")
    return frase == frase[::-1]

frase = input("Ingrese una frase: ")
if es_palindromo(frase):
    print("Es un palíndromo")
else:
    print("No es un palíndromo")
```

6. Realiza un algoritmo que cuente la cantidad de ocurrencias de cada caracter en una frase

```python
def contar_ocurrencias(frase):
    conteo = {}
    for caracter in frase:
        if caracter in conteo:
            conteo[caracter] += 1
        else:
            conteo[caracter] = 1
    return conteo

frase = input("Ingrese una frase: ")
print(contar_ocurrencias(frase))
```

7. Realiza un algoritmo que tome como entrada 3 puntos x,y del plano cartesiano, formando un triangulo, como salida el programa calcule al area y perimetro

```python
import math

x1 = float(input("x1: "))
y1 = float(input("y1: "))
x2 = float(input("x2: "))
y2 = float(input("y2: "))
x3 = float(input("x3: "))
y3 = float(input("y3: "))

lado1 = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
lado2 = math.sqrt((x3 - x2) ** 2 + (y3 - y2) ** 2)
lado3 = math.sqrt((x1 - x3) ** 2 + (y1 - y3) ** 2)

perimetro = lado1 + lado2 + lado3
area = abs((x1 * (y2 - y3) + x2 * (y3 - y1) + x3 * (y1 - y2)) / 2)

print(f"Perímetro: {perimetro:4.2f}")
print(f"Área: {area}")
```

8. Realiza un algoritmo que valide si un email es valido


```python
def es_email_valido(email):
    caracteresValidos = "abcdefghijklmnopqrstuvwxyzABCDEFGHIJKLMNOPQRSTUVWXYZ0123456789"

    partes = email.split("@")
    if len(partes) != 2:
        return False

    usuario = partes[0]
    dominio = partes[1]

    if len(usuario) == 0:
        return False
    for caracter in usuario:
        if caracter not in caracteresValidos:
            return False

    partesDominio = dominio.split(".")
    if len(partesDominio) < 2:
        return False

    extension = partesDominio[-1]
    if len(extension) < 2:
        return False
    for caracter in extension:
        if caracter not in caracteresValidos:
            return False

    return True

email = input("Ingrese un email: ")
if es_email_valido(email):
    print("Email válido")
else:
    print("Email inválido")
```

9. Compresor y Descompresor de Cadenas por RLE con Anidamiento Sencillo

```python
def comprimir_rle(cadena):
    if len(cadena) == 0:
        return ""
    resultado = ""
    caracterActual = cadena[0]
    contador = 1
    for i in range(1, len(cadena)):
        if cadena[i] == caracterActual:
            contador += 1
        else:
            resultado += caracterActual + str(contador)
            caracterActual = cadena[i]
            contador = 1
    resultado += caracterActual + str(contador)
    return resultado

def descomprimir_rle(cadena):
    digitos = "0123456789"
    resultado = ""
    i = 0
    while i < len(cadena):
        caracter = cadena[i]
        i += 1
        numero = ""
        while i < len(cadena) and cadena[i] in digitos:
            numero += cadena[i]
            i += 1
        resultado += caracter * int(numero)
    return resultado

original = "AAABBCDDDD"
comprimida = comprimir_rle(original)
descomprimida = descomprimir_rle(comprimida)

print(f"Original: {original}")
print(f"Comprimida: {comprimida}")
print(f"Descomprimida: {descomprimida}")
```

10. Conjetura de Collatz

```python
import matplotlib.pyplot as plt

def collatz(n):
    secuencia = [n]
    while n != 1:
        if n % 2 == 0:
            n = n // 2
        else:
            n = 3 * n + 1
        secuencia.append(n)
    return secuencia

numero = int(input("Ingrese un número entero positivo: "))
secuencia = collatz(numero)

pasos = range(0, len(secuencia))

plt.suptitle("Conjetura de Collatz")
plt.xlabel("Paso")
plt.ylabel("Valor")
plt.plot(pasos, secuencia)
plt.show()
```
